## Reduction Operations

```python
e = torch.randint(size = (2, 3), low = 0, high = 20, dtype = torch.float32)
e
```

```python
# sum
torch.sum(e)
# sum along columns
torch.sum(e, dim=0)
# sum along rows
torch.sum(e, dim=1)
```

```python
torch.mean(e)
torch.median(e, dim = 0)
torch.median(e, dim = 1)
```

```python
torch.max(e)
torch.min(e)
```

```python
torch.prod(e)
```

```python
torch.std(e)
```

```python
torch.var(e)
```

```python
torch.argmax(e)
```

```python
torch.argmin(e)
```

## Matrix Operations

```python
f = torch.randint(size=(2,3), low=0, high=10)
g = torch.randint(size=(3,2), low=0, high=10)

print(f)
print(g)
```

```python
# matrix multiplcation
torch.matmul(f, g)
```

```python
vector1 = torch.tensor([1, 2])
vector2 = torch.tensor([3, 4])

# dot product
torch.dot(vector1, vector2)
```

```python
# transpose
torch.transpose(f, 0, 1)
```

```python
h = torch.randint(size=(3,3), low=0, high=10, dtype=torch.float32)
torch.det(h)
```

```python
# inverse
torch.inverse(h)
```

## Comparision Oprations

```python
i = torch.randint(size=(2,3), low=0, high=10)
j = torch.randint(size=(2,3), low=0, high=10)

print(i)
print(j)
```

```python
# greater than
i > j
# less than
i < j
# equal to
i == j
# not equal to
i != j
```

## Special Functions

```python
k = torch.randint(size=(2,3), low=0, high=10, dtype=torch.float32)
k
```

```python
# log
torch.log(k)
# exp
torch.exp(k)
# sqrt
torch.sqrt(k)
# sigmoid
torch.sigmoid(k)
# softmax
torch.softmax(k, dim=0)
# relu
torch.relu(k)
```
